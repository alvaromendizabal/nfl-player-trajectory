# Observed-context supervision: matched confirmation study

Public, aggregate-only rendering of the latest AWS experiment. It contains no raw NFL tracking data, fitted weights, private player outputs, or cloud credentials.

**Outcome:** Fold 0 passed discovery, Fold 1 failed confirmation, so the family was not promoted.

```python
import pandas as pd
import plotly.graph_objects as go
from IPython.display import display

folds = pd.DataFrame([
    {"fold": 0, "reference": 0.4537257476309679,
     "control": 0.4530424213856637, "context": 0.45214906171205543},
    {"fold": 1, "reference": 0.47584636675081055,
     "control": 0.47618227735893615, "context": 0.47608927722206384},
])
display(folds)
```

```python
fig = go.Figure()
fig.add_bar(x=folds["fold"].astype(str), y=folds["reference"], name="Reference")
fig.add_bar(x=folds["fold"].astype(str), y=folds["control"], name="Matched control")
fig.add_bar(x=folds["fold"].astype(str), y=folds["context"], name="Context treatment")
fig.update_layout(title="Grouped-fold coordinate RMSE", barmode="group",
                  xaxis_title="Fold", yaxis_title="Coordinate RMSE (yards)",
                  width=900, height=500, template=None)
fig.show()
```

## Matched-control interpretation

Fold 0 improved the fixed blend by 0.001577 RMSE versus the reference and by 0.000893 versus matched fine-tuning. Fold 1 did not confirm the effect. The predeclared protocol stopped the family rather than tuning weights or selecting the favorable fold.

```python
effects = pd.DataFrame([
    {"fold": "0", "reference_gain": 0.0015766859189124682,
     "incremental_gain": 0.0008933596736082627},
    {"fold": "1", "reference_gain": -0.00024291047125329923,
     "incremental_gain": 0.00009300013687230102},
])
fig = go.Figure()
fig.add_bar(x=effects["fold"], y=effects["reference_gain"], name="Gain vs reference")
fig.add_bar(x=effects["fold"], y=effects["incremental_gain"],
            name="Incremental vs matched control")
fig.add_hline(y=0, line_dash="dash")
fig.update_layout(title="Predeclared effect sizes", barmode="group",
                  xaxis_title="Fold", yaxis_title="RMSE reduction",
                  width=900, height=500, template=None)
fig.show()
```
